# Explicación del `__init__`

```python
# dunder -> double underscore

# __new__
# __add__
# __init__
```

```python
class MyCustomScaler:

    def __init__(self, verbose):
        self.verbose = verbose

    def fit(self, X):

        mean_ = X.mean()
        var_ = X.var(ddof=0)

        if self.verbose:
            print(mean_)
            print(var_)

        self.mean_ = mean_
        self.var_ = var_

        return self

    def transform(self, X):

        Xt = (X - self.mean_) / (self.var_**0.5)

        return Xt
```

```python
my_scaler = MyCustomScaler(verbose=False)
```

```python
my_scaler.fit(X)
```

# Herencia con dummy Mixin

```python
class MyTransformerMixin:

    def fit_transform(self, X):

        print("Hello from MyTransformerMixin")
        Xt = self.fit(X=X).transform(X=X)

        return Xt
```

```python
class MyCustomScaler2(MyTransformerMixin):

    def __init__(self, verbose):
        self.verbose = verbose

    def fit(self, X):

        mean_ = X.mean()
        var_ = X.var(ddof=0)

        if self.verbose:
            print(mean_)
            print(var_)

        self.mean_ = mean_
        self.var_ = var_

        return self

    def transform(self, X):

        Xt = (X - self.mean_) / (self.var_**0.5)

        return Xt
```

```python
my_scaler2 = MyCustomScaler2(verbose=True)
```

```python
my_scaler2.fit_transform(X=X)
```

# Herencia con TransformerMixin y BaseEstimator

```python
from sklearn.base import TransformerMixin, BaseEstimator
```

```python
class MyCustomScaler(BaseEstimator, TransformerMixin):

    def __init__(self, verbose):
        self.verbose = verbose

    def fit(self, X):

        mean_ = X.mean()
        var_ = X.var(ddof=0)

        if self.verbose:
            print(mean_)
            print(var_)

        self.mean_ = mean_
        self.var_ = var_

        return self

    def transform(self, X):

        Xt = (X - self.mean_) / (self.var_**0.5)

        return Xt
```

```python
my_scaler = MyCustomScaler(verbose=True)
```
